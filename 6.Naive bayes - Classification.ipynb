```python
#Importing the Libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Reading the dataset
dataset=pd.read_csv("Social_Network_Ads.csv") #Dataset file name
dataset #Function call
```

```python
dataset=pd.get_dummies(dataset,drop_first=True) #Categorical data converted into numerical data. This technique is called One-Hot Encoding.
```

```python
dataset
```

```python
dataset=dataset.drop("User ID",axis=1) #Drop - Remove the user ID
dataset
```

```python
dataset["Purchased"].value_counts() #To Identify Balanced or Imbalanced Value Counts
```

```python
indep=dataset[["Age","EstimatedSalary","Gender_Male"]] #Independent - Input dataset
dep=dataset["Purchased"] #Dependent - Output dataset
```

```python
indep.shape
```

```python
dep
```

```python
#split into training set and test set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(indep, dep, test_size = 1/3, random_state = 0)
```

```python
#Model creation
from sklearn.naive_bayes import GaussianNB  #These first two lines are called as a procedure
classifier =GaussianNB() 
classifier.fit(X_train, y_train) #Fit - Called Function. X-input, Y-output
```

```python
y_pred = classifier.predict(X_test) #Evaluation Metric - For Model Evaluation
```

```python
from sklearn.metrics import confusion_matrix #Confusion Matrix Identification
cm = confusion_matrix(y_test, y_pred)
        
```

```python
from sklearn.metrics import classification_report #Classification Report
clf_report = classification_report(y_test, y_pred)
```

```python
print(clf_report) #Function Call
```

```python
print(cm)
```
